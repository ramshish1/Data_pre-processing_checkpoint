```python
import pandas as pd
import numpy as np
from scipy import stats

# Create a DataFrame with sample data
df = pd.DataFrame({'Height': [170, 160, 130, 190, 180, 150, 140, 200, 175, 165,500]})

# Calculate the z-score of each data point
z_scores = np.abs(stats.zscore(df['Height']))
threshold = 3
outliers = df[z_scores > threshold]

print("Outliers based on z-score:")
print(outliers)
```

```python
# Calculate Q1, Q3, and IQR
Q1 = df['Height'].quantile(0.25)
Q3 = df['Height'].quantile(0.75)
IQR = Q3 - Q1

# Define threshold and detect outliers
threshold = 1.5
outliers = df[(df['Height'] < (Q1 - threshold * IQR)) | (df['Height'] > (Q3 + threshold * IQR))]

print("Outliers based on IQR:")
outliers
```

```python
df_clean = df.drop(outliers.index)
df_clean
```

```python
outliers.index
```

```python
# Replace outliers with the median value of the dataset
median_value = df['Height'].median()
df.loc[df.index.isin(outliers.index), 'Height'] = median_value
df
```

```python
import pandas as pd
# Create a sample dataframe
df = pd.DataFrame({'Priority': ['Low', 'Medium', 'High', 'Medium', 'Low']})
df
```

```python
# Define a mapping dictionary
priority_mapping = {'Low': 1, 'Medium': 2, 'High': 3}
# Use the replace function to apply the dictionary mapping
df['Priority_Encoded'] = df['Priority'].map(priority_mapping)
# Display the updated dataframe
df
```
